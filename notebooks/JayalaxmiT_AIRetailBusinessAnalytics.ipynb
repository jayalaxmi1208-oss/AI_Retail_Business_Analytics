# AI-Driven Retail Business Analytics & Sales Prediction

## Project Overview

This project analyzes retail transaction data to identify sales,
profitability, customer, product and regional performance patterns.

The project combines:

- Data Cleaning
- Exploratory Data Analysis
- Business Analytics
- Key Performance Indicators (KPIs)
- Machine Learning
- Sales Prediction
- AI-Assisted Executive Analysis
- Interactive Business Dashboard

The objective is to transform raw retail transaction data into
verified business insights that can support management decision-making.

## 1. Business Problem

Retail organizations generate large amounts of transactional data,
but raw sales records do not immediately explain whether the business
is performing well, which products are profitable, where business
risks exist, or which areas may provide opportunities.

This project aims to analyze retail sales data and develop an
AI-assisted business analytics solution that helps answer:

1. How are sales and profitability performing?
2. Which products and categories contribute most to the business?
3. Which regions and customer segments perform differently?
4. Where are potential profitability risks?
5. How have sales changed over time?
6. What are the expected future sales trends?
7. What business insights can be generated from verified metrics?

## 2. Project Objectives

The main objectives are:

- Clean and validate the retail transaction dataset.
- Perform exploratory data analysis.
- Calculate executive-level business KPIs.
- Analyze sales, profit, products, customers and regions.
- Identify potential business risk indicators.
- Analyze relationships such as discount and profitability.
- Build a machine learning model for sales prediction.
- Generate AI-assisted executive insights using verified metrics.
- Develop a three-page business dashboard.
- Present findings in a clear business storytelling format.

## 3. Dataset Description

### Dataset Name
Superstore Dataset

### Source
Kaggle

### Purpose

The dataset contains retail transaction information including orders,
customers, products, categories, regions, sales, quantity, discount
