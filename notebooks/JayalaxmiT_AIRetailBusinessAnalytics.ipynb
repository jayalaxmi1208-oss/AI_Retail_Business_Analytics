## 1. Business Problem

Retail organizations generate large amounts of transactional data,
but raw sales records do not immediately explain whether the business
is performing well, which products are profitable, where business
risks exist, or which areas may provide opportunities.

This project aims to analyze retail sales data and develop an
AI-assisted business analytics solution that helps answer:

1. How are sales and profitability performing?
2. Which products and categories contribute most to the business?
3. Which regions and customer segments perform differently?
4. Where are potential profitability risks?
5. How have sales changed over time?
6. What are the expected future sales trends?
7. What business insights can be generated from verified metrics?

```python
clean_df = df.copy()

print("Clean working copy created.")
print("Original rows:", len(df))
print("Clean working copy rows:", len(clean_df))
```

```python
clean_df.columns = clean_df.columns.str.strip()

print("Column names standardized successfully.")
print(clean_df.columns.tolist())
```

```python
print("Date column data types:")
print(clean_df[["Order Date", "Ship Date"]].dtypes)
```

```python
# ==============================
# FINAL DATA CLEANING PREPARATION
# ==============================

# 1. Remove accidental spaces from text values
text_columns = clean_df.select_dtypes(include="object").columns

for column in text_columns:
    clean_df[column] = clean_df[column].str.strip()

# 2. Create useful date-based features
clean_df["Year"] = clean_df["Order Date"].dt.year
clean_df["Month"] = clean_df["Order Date"].dt.month
clean_df["Quarter"] = clean_df["Order Date"].dt.quarter
clean_df["Order Month"] = clean_df["Order Date"].dt.to_period("M").astype(str)

# 3. Calculate shipping duration
clean_df["Shipping Days"] = (
    clean_df["Ship Date"] - clean_df["Order Date"]
).dt.days

# 4. Verify the cleaned dataset
print("Cleaning and feature preparation completed!")
print("\nRows:", clean_df.shape[0])
print("Columns:", clean_df.shape[1])

print("\nMissing values:")
print(clean_df.isnull().sum().sum())

print("\nDuplicate rows:")
print(clean_df.duplicated().sum())

print("\nNew columns:")
print(["Year", "Month", "Quarter", "Order Month", "Shipping Days"])
```

## 6. Data Cleaning and Feature Preparation

A clean working copy of the original dataset was created so that the
raw dataset remained unchanged.

The following data preparation steps were performed:

- Standardized column names by removing unnecessary spaces.
- Removed leading and trailing spaces from text values.
- Verified that there were no missing values.
- Verified that there were no duplicate rows.
- Converted Order Date and Ship Date into datetime format.
- Verified that Sales, Quantity and Discount contained valid values.
- Retained negative Profit values because they represent legitimate
  business losses rather than invalid data.
- Created Year, Month and Quarter features for time-based analysis.
- Created Order Month for monthly sales analysis and forecasting.
- Created Shipping Days to analyze delivery duration.

After cleaning and feature preparation, the dataset contained
9,994 rows and 26 columns.

```python
monthly_sales = (
    clean_df.groupby("Order Month")["Sales"]
    .sum()
    .reset_index()
)

monthly_sales["Order Month"] = pd.to_datetime(monthly_sales["Order Month"])

print("Monthly sales data:")
print(monthly_sales.head())
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales["Order Month"],
    monthly_sales["Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Order Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
monthly_profit = (
    clean_df.groupby("Order Month")["Profit"]
    .sum()
    .reset_index()
)

monthly_profit["Order Month"] = pd.to_datetime(
    monthly_profit["Order Month"]
)

print("Monthly profit data:")
print(monthly_profit.head())
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_profit["Order Month"],
    monthly_profit["Profit"],
    marker="o"
)

plt.title("Monthly Profit Trend")
plt.xlabel("Order Month")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
category_performance = (
    clean_df.groupby("Category")[["Sales", "Profit"]]
    .sum()
    .reset_index()
)

print("Category Performance:")
print(category_performance)
```

```python
plt.figure(figsize=(10, 6))

x = range(len(category_performance))

plt.bar(
    [i - 0.2 for i in x],
    category_performance["Sales"],
    width=0.4,
    label="Sales"
)

plt.bar(
    [i + 0.2 for i in x],
    category_performance["Profit"],
    width=0.4,
    label="Profit"
)

plt.xticks(x, category_performance["Category"])
plt.title("Sales and Profit by Category")
plt.xlabel("Category")
plt.ylabel("Amount")
plt.legend()
plt.tight_layout()

plt.show()
```

```python
region_performance = (
    clean_df.groupby("Region")[["Sales", "Profit"]]
    .sum()
    .reset_index()
)

print("Regional Performance:")
print(region_performance)
```

```python
plt.figure(figsize=(10, 6))

x = range(len(region_performance))

plt.bar(
    [i - 0.2 for i in x],
    region_performance["Sales"],
    width=0.4,
    label="Sales"
)

plt.bar(
    [i + 0.2 for i in x],
    region_performance["Profit"],
    width=0.4,
    label="Profit"
)

plt.xticks(x, region_performance["Region"])
plt.title("Sales and Profit by Region")
plt.xlabel("Region")
plt.ylabel("Amount")
plt.legend()
plt.tight_layout()

plt.show()
```

```python
# Calculate Average Order Value (AOV)

total_sales = clean_df["Sales"].sum()
total_orders = clean_df["Order ID"].nunique()

average_order_value = total_sales / total_orders

print("Total Sales:", round(total_sales, 2))
print("Total Orders:", total_orders)
print("Average Order Value:", round(average_order_value, 2))
```

```python
# Calculate Profit Margin

total_sales = clean_df["Sales"].sum()
total_profit = clean_df["Profit"].sum()

profit_margin = (total_profit / total_sales) * 100

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Profit Margin:", round(profit_margin, 2), "%")
```

```python
# Calculate Year-over-Year Sales Growth

yearly_sales = (
    clean_df.groupby("Year")["Sales"]
    .sum()
    .reset_index()
)

yearly_sales["Sales Growth %"] = (
    yearly_sales["Sales"].pct_change() * 100
)

print("Yearly Sales and Growth:")
print(yearly_sales.round(2))
```

```python
# Calculate Total Revenue

total_revenue = clean_df["Sales"].sum()

print("Total Revenue:", round(total_revenue, 2))
```

```python
# Calculate Total Customers

total_customers = clean_df["Customer ID"].nunique()

print("Total Customers:", total_customers)
```

```python
# Calculate Total Orders

total_orders = clean_df["Order ID"].nunique()

print("Total Orders:", total_orders)
```

```python
# ==========================================
# EXECUTIVE BUSINESS KPIs
# ==========================================

# Total Revenue / Sales
total_revenue = clean_df["Sales"].sum()

# Total Profit
total_profit = clean_df["Profit"].sum()

# Total Orders
total_orders = clean_df["Order ID"].nunique()

# Total Customers
total_customers = clean_df["Customer ID"].nunique()

# Average Order Value
average_order_value = total_revenue / total_orders

# Profit Margin
profit_margin = (total_profit / total_revenue) * 100

# Year-over-Year Sales Growth
yearly_sales = (
    clean_df.groupby("Year")["Sales"]
    .sum()
    .reset_index()
)

yearly_sales["Sales Growth %"] = (
    yearly_sales["Sales"].pct_change() * 100
)

# Latest available year's growth
sales_growth = yearly_sales["Sales Growth %"].iloc[-1]

# Display all KPIs
print("========== EXECUTIVE BUSINESS KPIs ==========")
print(f"Total Revenue:       {total_revenue:,.2f}")
print(f"Total Profit:        {total_profit:,.2f}")
print(f"Total Orders:        {total_orders:,}")
print(f"Total Customers:     {total_customers:,}")
print(f"Average Order Value: {average_order_value:,.2f}")
print(f"Profit Margin:       {profit_margin:.2f}%")
print(f"Sales Growth:        {sales_growth:.2f}%")
```

```python
# ==========================================
# KPI 1 — AVERAGE PROFIT PER ORDER
# ==========================================

average_profit_per_order = total_profit / total_orders

print("Average Profit per Order:", round(average_profit_per_order, 2))
```

```python
# ==========================================
# KPI 2 — TOTAL UNITS SOLD
# ==========================================

total_units_sold = clean_df["Quantity"].sum()

print("Total Units Sold:", total_units_sold)
```

```python
# ==========================================
# KPI 3 — AVERAGE UNITS PER ORDER
# ==========================================

average_units_per_order = total_units_sold / total_orders

print("Average Units per Order:", round(average_units_per_order, 2))
```

```python
# ==========================================
# KPI 4 — LOSS-MAKING ORDERS
# ==========================================

order_profit = (
    clean_df.groupby("Order ID")["Profit"]
    .sum()
    .reset_index()
)

loss_making_orders = (order_profit["Profit"] < 0).sum()

loss_making_order_percentage = (
    loss_making_orders / total_orders
) * 100

print("Loss-Making Orders:", loss_making_orders)
print(
    "Loss-Making Order Percentage:",
    round(loss_making_order_percentage, 2),
    "%"
)
```

```python
# ==========================================
# CHART 1 — PROFITABLE VS LOSS-MAKING ORDERS
# ==========================================

profitable_orders = (order_profit["Profit"] >= 0).sum()

order_profit_status = pd.DataFrame({
    "Order Status": ["Profitable Orders", "Loss-Making Orders"],
    "Orders": [profitable_orders, loss_making_orders]
})

print(order_profit_status)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    order_profit_status["Order Status"],
    order_profit_status["Orders"]
)

plt.title("Profitable vs Loss-Making Orders")
plt.xlabel("Order Status")
plt.ylabel("Number of Orders")

plt.tight_layout()
plt.show()
```
